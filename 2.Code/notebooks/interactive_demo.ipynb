# Interactive CC-MPC Simulation

Run the CC-MPC algorithm with matplotlib visualization — no MuJoCo required.

Modify the parameters below and re-run cells to explore different scenarios.

```python
import sys
sys.path.insert(0, '..')

from simulation.runner import SimulationRunner
from simulation.visualizer import MatplotlibVisualizer
```

## Configuration

Select a scenario and adjust parameters:

```python
CONFIG_PATH = "config/scenarios/two_static.yaml"
SEED = 42

# Create runner and run simulation
runner = SimulationRunner(CONFIG_PATH)
history = runner.run(seed=SEED)
summary = runner._make_summary(history, SEED)
print(summary)
```

## Visualize Trajectory

Generate the 2D trajectory plot inline:

```python
import matplotlib.pyplot as plt
%matplotlib inline

viz = MatplotlibVisualizer(history, dpi=100)
viz.plot_trajectory("output/trajectory.png")
plt.imshow(plt.imread("output/trajectory.png"))
plt.axis('off')
plt.show()
```

## Summary Panel

Multi-panel figure showing trajectory, altitude, controls, and solve time:

```python
viz.plot_summary_panel("output/summary.png")
plt.imshow(plt.imread("output/summary.png"))
plt.axis('off')
plt.show()
```
